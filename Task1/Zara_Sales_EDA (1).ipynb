## 6. Outlier Detection

We use the IQR method and Z-score to detect and understand outliers in Price and Sales Volume.

```python
#IQR method: identify outliers in Price and Sales Volume
def detect_outliers_iqr(series, col_name):
    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    outliers = series[(series < lower) | (series > upper)]
    print(f"\n── {col_name} Outliers (IQR Method) ──")
    print(f"  Q1={Q1:.2f}, Q3={Q3:.2f}, IQR={IQR:.2f}")
    print(f"  Lower bound: {lower:.2f} | Upper bound: {upper:.2f}")
    print(f"  Number of outliers: {len(outliers)} ({len(outliers)/len(series)*100:.1f}%)")
    return lower, upper

price_lower, price_upper = detect_outliers_iqr(df['price'], 'Price')
vol_lower, vol_upper = detect_outliers_iqr(df['Sales Volume'], 'Sales Volume')
```

```python
#Visualize outliers with annotated boxplots
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.boxplot(y=df['price'], ax=axes[0], color='steelblue', flierprops=dict(marker='o', color='red', alpha=0.5))
axes[0].set_title('Price — Outliers Highlighted')
axes[0].axhline(price_upper, color='red', linestyle='--', linewidth=1, label=f'Upper bound ({price_upper:.0f})')
axes[0].legend(fontsize=8)

sns.boxplot(y=df['Sales Volume'], ax=axes[1], color='coral', flierprops=dict(marker='o', color='red', alpha=0.5))
axes[1].set_title('Sales Volume — Outliers Highlighted')
axes[1].axhline(vol_upper, color='red', linestyle='--', linewidth=1, label=f'Upper bound ({vol_upper:.0f})')
axes[1].legend(fontsize=8)

plt.suptitle('Outlier Detection — IQR Boundaries', fontsize=13, fontweight='bold')
plt.show()
```

```python
#Sales Volume vs. Product Position
#Does catalog/store position affect how many units are sold?
if 'Product Position' in df.columns:
    fig, ax = plt.subplots(figsize=(10, 5))
    ax.scatter(df['Product Position'], df['Sales Volume'], alpha=0.4, color='mediumpurple', s=20)
    ax.set_title('Sales Volume vs. Product Position')
    ax.set_xlabel('Product Position')
    ax.set_ylabel('Sales Volume')
    plt.show()
```

```python
#Promoted vs. Non-Promoted: Revenue comparison
promo_rev = df.groupby('Promotion')['Revenue'].sum()

fig, ax = plt.subplots(figsize=(7, 5))
ax.bar(promo_rev.index.astype(str), promo_rev.values,
       color=['steelblue', 'coral'], edgecolor='white', width=0.5)
ax.set_title('Total Estimated Revenue: Promoted vs. Non-Promoted')
ax.set_xlabel('On Promotion')
ax.set_ylabel('Total Revenue')
for i, val in enumerate(promo_rev.values):
    ax.text(i, val * 1.01, f'{val:,.0f}', ha='center', fontsize=10)
plt.tight_layout()
plt.savefig('revenue_promo_vs_not.png', dpi=150)
plt.show()
```

#### 1. Dataset Structure
The Zara sales dataset contains product-level data across multiple categories with 17 features including price, sales volume, promotion status, and seasonality flags. The dataset required minor cleaning, but was generally of good quality.

---

#### 2. Pricing Patterns
- Price distribution is **right-skewed**: most products fall in a lower price range, with a tail of premium items driving the mean above the median.
- **Promoted items** tend to have lower average prices, consistent with discounting being used as a promotional mechanism.

---

#### 3. Sales Volume Patterns
- Sales volume is also right-skewed — a small number of bestsellers account for a disproportionate share of units sold.
- Categories with lower prices tend to show higher average sales volumes, suggesting **price elasticity** is present in Zara's market.
- **Promoted products** tend to sell more units, confirming that promotions are effective at driving volume.

---

#### 4. Revenue Insights
- Estimated revenue (Price × Sales Volume) reveals that **high-volume mid-priced items** often generate more total revenue than premium-priced low-volume products.
- A small number of products (top 10) likely contribute significantly to total revenue — a classic **Pareto distribution** pattern.

---

#### 5. Promotions & Seasonality
- Promotions appear to increase sales volume but reduce unit price, creating a trade-off between margin and volume.
- Seasonal products may occupy different price tiers than year-round products, reflecting limited-edition or trend-driven pricing.

---

#### 6. Outliers
- Outliers in price represent legitimate premium products (not data errors) and should be kept in the dataset.
- Outliers in sales volume represent viral or bestseller products that are strategically important for the business.

---

#### 7. Recommendations for the Business

| Insight | Business Recommendation |
|---|---|
| Promotions drive volume but reduce price | Use promotions strategically for slow-moving stock, not bestsellers |
| Small number of products drive majority of revenue | Focus inventory and marketing on top-performing SKUs |

| Seasonal items have distinct pricing | Plan seasonal collections with dedicated pricing strategies |

#### Conclusion
This EDA reveals that Zara's sales data follows patterns typical of fashion retail: skewed price and volume distributions, significant category-driven price variation, and clear promotion-driven volume boosts.
